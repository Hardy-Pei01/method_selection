# Convergence diagnostic — all benchmarks

Convergence checking now lives in `analysis_lib.diagnose_convergence`. You call
it **once per benchmark**, passing a list of sections (one per paradigm/setting)
with the folder regex and `condition_format` for that benchmark — the same ones
already defined in each analysis notebook. The logic is shared, so tree, lake,
and constrained-lake all use the identical diagnostic.

**Reading it**
- **MOEA**: watch `gain_last20pct` (share of `epsilon_progress` gained in the
  last 20% of NFE). <2% = converged, 2–5% = borderline, ≥5% = under-budgeted.
- **MORL**: `verdict == converged` means every seed drove `pcs_shift_eps` to ~0
  by `conv_step`; otherwise `final_shift` shows how far it stayed from zero.

Missing directories are skipped (not fatal), so partial data is fine.

```python
import os, re
import analysis_lib as A
```

## Tree  *(validated)*

```python
ROOT = '../data/tree_data_2'

MOEA_DET = re.compile(r'^(intertemporal|table)_(NSGAII|IBEA|MOEAD)_single_(\d+)_(observable|non_observable)$')
MORL_DET = re.compile(r'^(pareto|indicator|decomposition)_single_(\d+)$')
MOEA_ROB = re.compile(r'^(intertemporal|table)_(NSGAII|IBEA|MOEAD)_(multi|moro)_(\d+)_observable$')
MORL_ROB = re.compile(r'^(pareto|indicator|decomposition)_(multi|moro)_(\d+)$')

A.diagnose_convergence([
    {'label':'Deterministic MOEA','kind':'moea','base':f'{ROOT}/deterministic/moea',
     'folder_re':MOEA_DET, 'fmt':lambda g:(f'{g[0]}_{g[3]}', g[1], int(g[2]))},
    {'label':'Robust MOEA',       'kind':'moea','base':f'{ROOT}/robust/moea',
     'folder_re':MOEA_ROB, 'fmt':lambda g:(f'{g[0]}_{g[2]}', g[1], int(g[3]))},
    {'label':'Deterministic MORL','kind':'morl','base':f'{ROOT}/deterministic/morl',
     'folder_re':MORL_DET, 'fmt':lambda g:('closed_loop_single', g[0], int(g[1]))},
    {'label':'Robust MORL',       'kind':'morl','base':f'{ROOT}/robust/morl',
     'folder_re':MORL_ROB, 'fmt':lambda g:(f'closed_loop_{g[1]}', g[0], int(g[2]))},
], title='TREE')
```
